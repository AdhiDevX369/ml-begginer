```python
# lets do dot product of two vectors
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
dot_product = np.dot(a, b)

# print every step of the calculation
print(f"Vector a: {a}")
print(f"Vector b: {b}")
print(f"Dot product calculation: {a[0]}*{b[0]} + {a[1]}*{b[1]} + {a[2]}*{b[2]} = {dot_product}")
```
